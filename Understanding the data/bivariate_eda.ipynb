

### The Anatomy of a Boxplot

Here is exactly what each part of the visualization represents:

* **The Center Line (Median):** The line inside the box represents the **Median** (). This is the middle value of the data—50% of the data points fall below this line, and 50% fall above.
* **The Box (Interquartile Range - IQR):** The box itself represents the middle 50% of your data.
* **Bottom of the box:** The 25th percentile ().
* **Top of the box:** The 75th percentile ().
* The height of the box is the **IQR** ().


* **The Whiskers:** These lines extend from the box to show the rest of the distribution.
* By default, Seaborn extends whiskers to  beyond the quartiles.
* They represent the minimum and maximum values that are **not** considered outliers.


* **Individual Dots (Outliers):** Any data points that fall outside the whiskers are plotted as individual points (often called "fliers"). These are mathematically determined "extreme" values.
