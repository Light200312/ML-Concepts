# Encoding
Encoding is the process of converting categorical data into a numerical format that can be used by machine learning algorithms because models cannot directly process categorical data. 
types of data that can be encoded include:
- Nominal data: Categorical data without any inherent order (e.g., colors, brands).
- Ordinal data: Categorical data with a clear order or ranking (e.g., education levels, customer satisfaction ratings).

## types of encoding include:
- **Ordinal Encoding**: Assigns a unique integer to each category based on their order. It is suitable for ordinal data where the categories have a meaningful order.It is used for input columns.
- **Label Encoding**: Similar to ordinal encoding but does not consider the order of categories. It assigns a unique integer to each category without any specific order. It is used for output column only. For we can also use One-Hot-Encoding.
- **One-Hot Encoding**: Creates binary columns for each category, indicating the presence or absence of that category. It is suitable for nominal data where there is no inherent order. It is used for input columns. To counter the dummy variable trap(dependency in input variable- Must be avoided), we can drop one of the binary columns created by one-hot encoding.The dropped column serves as a reference category, and the remaining columns indicate the presence of the other categories relative to the reference category. 00 indicated droped column.




## Step 1: Import Required Libraries
Import the necessary libraries for data preprocessing and encoding:
- **pandas**: For data manipulation and creating DataFrames
- **numpy**: For numerical operations
- **sklearn.model_selection**: For splitting data into train-test sets
- **sklearn.preprocessing**: For OrdinalEncoder (features) and LabelEncoder (target)

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OrdinalEncoder, LabelEncoder
```

## Step 2: Load Data from CSV File
Read the customer data from the CSV file and display the first few rows to understand the structure of the data.

```python
df= pd.read_csv("customer.csv")
df.head()
```

## Step 3: Drop Irrelevant Columns
Remove the first two columns (index columns) keeping only the relevant features for analysis.

```python
df=df.iloc[:,2:]
df.head()
```

## Step 4: Separate Features and Target Variable
Split the data into:
- **Features (X)**: Independent variables used for prediction (Review, Education)
- **Target (y)**: The dependent variable we want to predict (Purchased)

```python
x=df.drop("purchased",axis=1)
y=df["purchased"]
x,y
```

## Step 5: Train-Test Split
Divide the data into training and testing sets:
- **Training set (80%)**: Used to train the encoding models
- **Testing set (20%)**: Used to evaluate the model performance
- **random_state=42**: Ensures reproducibility of the split

```python
X_train,X_test,Y_train,Y_test=train_test_split(x,y,test_size=0.2,random_state=42)
X_train.head(5)
```

## Step 6: Display First 5 Training Samples
View the first 5 rows of the training features to understand the data structure.

```python
Y_train.head(5)
```

## Step 7: Display Training Target Values
Show the first 5 values of the training target variable (before encoding).

```python
ordinal_encoder=OrdinalEncoder(categories=[['Poor','Average','Good'],['School','UG','PG']])
X_train_encoded=ordinal_encoder.fit_transform(X_train)
ordinal_encoder.categories_
```

## Step 8: Ordinal Encoding for Features
Apply Ordinal Encoding to convert categorical features into numerical values:
- Creates a new integer for each category
- **fit_transform()**: Learns the categories and encodes them simultaneously
- **categories_**: Shows the unique categories found for each feature

```python
X_train_encoded
```

## Step 9: Display Encoded Features
View the encoded feature values as a numpy array.

```python
le=LabelEncoder()
Y_train_encoded=le.fit_transform(Y_train)
Y_train_encoded
```

```python
le.classes_
```

## Step 10: Label Encoding for Target Variable
Apply Label Encoding to convert the target variable into numerical format:
- Used only for output/target columns
- Assigns unique integers to each category
- **fit_transform()**: Learns and encodes the target variable

```python
x_df=pd.DataFrame(X_train_encoded,columns=['Review','Education'])
y_df=pd.DataFrame(Y_train_encoded,columns=['Purchased'])
x_df.head(5)
```

## Step 11: Convert Encoded Features to DataFrame
Convert the numpy arrays into pandas DataFrames with proper column names for better data manipulation:
- **x_df**: Contains encoded features (Review, Education)
- **y_df**: Contains encoded target variable (Purchased)

```python
y_df.head()
```

## Step 12: Display Encoded Target DataFrame
View the first few rows of the encoded target variable.

```python
df_encoded=pd.concat([x_df,y_df],axis=1)# axis=1 means we are concatenating column wise, axis=0 means we are concatenating row wise
df_encoded.head()
```

## Step 13: Combine Encoded Features and Target
Merge the encoded features and encoded target variable into a single DataFrame using **pd.concat()**:
- **axis=1**: Combines columns side-by-side (horizontally)
- **Result**: A complete dataset with encoded features and target for machine learning

# ** One-Hot Encoding** 
One-Hot Encoding is a technique used to convert categorical variables into a format that can be provided to machine learning algorithms to improve predictions. It creates binary columns for each category, indicating the presence or absence of that category. It is suitable for nominal data where there is no inherent order. It is used for input columns. To counter the dummy variable trap(dependency in input variable- Must be avoided), we can drop one of the binary columns created by one-hot encoding.The dropped column serves as a reference category, and the remaining columns indicate the presence of the other categories relative to the reference category. 00 indicated droped column.

```python
from sklearn.preprocessing import OneHotEncoder
ohe=OneHotEncoder(drop='first',sparse_output=False) # drop='first' is used to drop the first column created by one-hot encoding to counter the dummy variable trap, sparse=False is used to return a dense array instead of a sparse matrix
```
